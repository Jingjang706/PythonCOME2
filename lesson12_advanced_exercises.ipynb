## Exercise 1: Robust numeric parser

- Write a function that parses user-provided string to int or float.
- Raise/handle exceptions for invalid formats.
- Return tuple `(success, value_or_error)` and print examples.

```python
def parse_number(text):
    try:
        return (True, int(text))
    except (ValueError, TypeError):
        pass
    try:
        return (True, float(text))
    except (ValueError, TypeError):
        return (False, f"Invalid number: {text!r}")

for sample in ["42", "3.14", "-7", "abc", "", None]:
    print(sample, "->", parse_number(sample))
```

## Exercise 2: Multi-exception workflow

- Build a division function that converts both inputs with `float()` and divides, using `try/except/else/finally`.
- Handle `TypeError`, `ValueError`, and `ZeroDivisionError` separately.
- Demonstrate all branches with different calls.

```python
def safe_divide(a, b):
    try:
        result = float(a) / float(b)
    except TypeError:
        print("TypeError: inputs must be numbers or numeric strings.")
    except ValueError:
        print("ValueError: could not convert input to a number.")
    except ZeroDivisionError:
        print("ZeroDivisionError: cannot divide by zero.")
    else:
        print(f"Result: {result}")
    finally:
        print("Division attempt finished.\n")

safe_divide(10, 2)
safe_divide(10, 0)
safe_divide("abc", 2)
safe_divide(None, 2)
```

## Exercise 3: None-aware processing

- Given list with values and `None`, compute average of valid numbers only.
- Ignore `None` safely.
- Print cleaned list and average.

```python
data = [10, None, 20, None, 30, 40]
cleaned = [x for x in data if x is not None]
average = sum(cleaned) / len(cleaned) if cleaned else None
print("Cleaned list:", cleaned)
print("Average:", average)
```

## Exercise 4: Input validation loop

- Prompt repeatedly for integer between 1 and 100 until valid.
- Use exception handling and custom error messages.
- Print accepted value and attempt count.

```python
attempts = 0
while True:
    attempts += 1
    raw = input("Enter an integer between 1 and 100: ")
    try:
        value = int(raw)
    except ValueError:
        print("Invalid input: please enter a whole number.")
        continue
    if 1 <= value <= 100:
        break
    print("Out of range: the number must be between 1 and 100.")
print("Accepted value:", value)
print("Attempts:", attempts)
```

## Exercise 5: Mini CLI app

- Create text menu app (add/subtract/multiply/divide/quit).
- Use input + try/except for invalid choices and values.
- Keep looping until quit command.

```python
while True:
    print("\n1. Add  2. Subtract  3. Multiply  4. Divide  5. Quit")
    choice = input("Choose an option: ").strip().lower()
    if choice in ("5", "q", "quit"):
        print("Goodbye!")
        break
    if choice not in ("1", "2", "3", "4"):
        print("Invalid choice. Please pick 1-5.")
        continue
    try:
        a = float(input("First number: "))
        b = float(input("Second number: "))
        if choice == "1":
            print("Result:", a + b)
        elif choice == "2":
            print("Result:", a - b)
        elif choice == "3":
            print("Result:", a * b)
        else:
            print("Result:", a / b)
    except ValueError:
        print("Invalid number. Try again.")
    except ZeroDivisionError:
        print("Cannot divide by zero.")
```
