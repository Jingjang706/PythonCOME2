## Exercise 2: Iterator protocol deep dive

- Create iterator from a string and consume values with `next()` until exhausted.
- Handle `StopIteration` explicitly.
- Print total number of consumed characters.

```python
it = iter("Python")
count = 0
while True:
    try:
        ch = next(it)
    except StopIteration:
        print("Iterator exhausted")
        break
    print(ch)
    count += 1
print("Total consumed:", count)
```

## Exercise 3: Custom iterator class

- Implement class `EvenNumbers(limit)` yielding even numbers up to limit.
- Use `__iter__` and `__next__`.
- Iterate and print values.

```python
class EvenNumbers:
    def __init__(self, limit):
        self.limit = limit

    def __iter__(self):
        self.n = 0
        return self

    def __next__(self):
        self.n += 2
        if self.n > self.limit:
            raise StopIteration
        return self.n

for num in EvenNumbers(10):
    print(num)
```

## Exercise 4: Iterator state with `next()`

- Create a custom iterator class `Countdown(start)` that counts down to 1.
- Use `iter()` and `next()` to take the first two values manually.
- Then use a `for` loop on the same iterator to get the remaining values.
- Add a short comment explaining why the `for` loop continues where `next()` stopped.

```python
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < 1:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value

it = iter(Countdown(6))
print(next(it))
print(next(it))
for value in it:
    print(value)
# The iterator keeps its own state (self.current). iter(it) returns the same
# object, so the for loop resumes from where next() stopped.
```
